## 05_boxplot

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### используйте данные из ab-test, считайте их в фрейм данных и внесите любые изменения, которые могут пригодиться для решения задачи

```python
df = pd.read_csv("../data/ab-test.csv")
df
```

```python
df_before = df[df["time"] == "before"]
df_before.count()
```

```python
df_after = df[df["time"] == "after"]
df_after.count()
```

### размер шрифта остался прежним, вы можете выбрать любой размер шрифта, который вам нравится, цветовая палитра должна быть такой же, как в примере, размер шрифта заголовка равен 15, ширина линий интервалов равна 3, ширина медианных линий равна 2

```python
fig, ax = plt.subplots(1, 2, figsize=(15, 8), sharey=True)
df_after.boxplot(
    column="diff",
    by="group",
    ax=ax[0],
    showmeans=True,
    boxprops=dict(linewidth=3, color="green"),
    whiskerprops=dict(color="red"),
    capprops=dict(color="red"),
    medianprops=dict(linewidth=2, color="red"),
)
df_before.boxplot(
    column="diff",
    by="group",
    ax=ax[1],
    showmeans=True,
    boxprops=dict(linewidth=3, color="green"),
    whiskerprops=dict(color="red"),
    capprops=dict(color="red"),
    medianprops=dict(linewidth=2, color="red"),
)
ax[0].set_title("after")
ax[1].set_title("before")
fig.suptitle("Lag before labs deadlines. After and before the Newsfeed", fontsize=15)
```
